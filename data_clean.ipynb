```python
def collapse_binary_columns(
    df: pd.DataFrame,
    mapping: dict,
    new_col: str,
    default_value: str | None = None
):
    def get_value(row):
        active = [
            label
            for col, label in mapping.items()
            if col in row.index and row[col] == 1
        ]

        if len(active) == 1:
            return active[0]
        elif len(active) == 0:
            return default_value
        else:
            return "Multiple"   # data quality issue

    df[new_col] = df.apply(get_value, axis=1)

    # drop only columns that actually exist
    drop_cols = [c for c in mapping.keys() if c in df.columns]
    df.drop(columns=drop_cols, inplace=True)

    return df
```

```python
POSITION_COLS = {
    "Goalkeeper": "Goalkeeper",
    "Defender": "Defender",
    "Forward": "Forward"
}

df = collapse_binary_columns(
    df,
    mapping=POSITION_COLS,
    new_col="Position",
    default_value="Midfielder"
)
```

```python
R_CARD_COLS = {
    "r_card_1": "r_card_1",
    "r_card_2": "r_card_2",
    "r_card_3": "r_card_3"
}

df = collapse_binary_columns(
    df,
    mapping=R_CARD_COLS,
    new_col="r_card",
    default_value="r_card_0"
)
```

```python
TRANSFER_YEAR_COLS = {
    "Transfer_year_2011": "Transfer_year_2011",
    "Transfer_year_2012": "Transfer_year_2012",
    "Transfer_year_2013": "Transfer_year_2013",
    "Transfer_year_2014": "Transfer_year_2014",
    "Transfer_year_2015": "Transfer_year_2015",
    "Transfer_year_2016": "Transfer_year_2016",
    "Transfer_year_2017": "Transfer_year_2017",
    "Transfer_year_2018": "Transfer_year_2018",
    "Transfer_year_2019": "Transfer_year_2019",
    "Transfer_year_2020": "Transfer_year_2020",
    "Transfer_year_2021": "Transfer_year_2021",
    "Transfer_year_2022": "Transfer_year_2022",
    "Transfer_year_2023": "Transfer_year_2023",
    "Transfer_year_2024": "Transfer_year_2024",
    "Transfer_year_2025": "Transfer_year_2025",
    "Transfer_year_2026": "Transfer_year_2026",
}

df = collapse_binary_columns(
    df,
    mapping=TRANSFER_YEAR_COLS,
    new_col="Transfer_year",
    default_value="Transfer_year_2010"
)
```

```python
CONTINENT_COLS = {
    "Africa": "Africa",
    "Asia": "Asia",
    "Europe": "Europe",
    "North_America": "North_America",
    "South_America": "South_America"
}

df = collapse_binary_columns(
    df,
    mapping=CONTINENT_COLS,
    new_col="Continent",
    default_value="Oceania"
)
```

```python
df["Transfer_window"] = df["Winter_window"].map({
    0: "Summer_window",
    1: "Winter_window"
})

# drop old binary column
df.drop(columns=["Winter_window"], inplace=True)
```

```python
EUROPEAN_COMPETITIONS_COLS = {
    "champions_league": "champions_league",
    "europa_league": "europa_league"
}

df = collapse_binary_columns(
    df,
    mapping=EUROPEAN_COMPETITIONS_COLS,
    new_col="european_competitions",
    default_value="no_participation"
)
```

```python
BINARY_EXPLICIT_COLS = [
    "English",
    "Homegrown",
    "From_Premier_League",
    "From_top_5_leagues",
    "From_FA_leagues",
    "From_europe_leagues",
    "From_south_america_leagues",
    "From_big_six",
    "To_Premier_League",
    "To_top_5_leagues",
    "To_FA_leagues",
    "To_europe_leagues",
    "To_south_america_leagues",
    "To_saudi_arabic_league",
    "To_big_six",
]

def binary_to_explicit_category(df, columns):
    """
    Convert binary columns (0/1) into explicit categorical variables:
    1 -> column name
    0 -> 'not_' + column name
    """
    for col in columns:
        if col not in df.columns:
            continue

        df[col] = df[col].map({
            1: col,
            0: f"not_{col}"
        })

    return df

df = binary_to_explicit_category(df, BINARY_EXPLICIT_COLS)
```

```python
FOOT_COLS = {
    "is_right": "right",
    "is_both": "both"
}

df = collapse_binary_columns(
    df,
    mapping=FOOT_COLS,
    new_col="Foot",
    default_value="left"
)
```
